Data analisys to answer the following question:

What the data about mobile telephony reveal about the technologic evolution e distribution of access in Brazil?

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/br_anatel_indice_brasileiro_conectividade_municipio.csv')

df.shape
```

```python
print(df.info())

print(df.describe())

df.select_dtypes(include=['number']).corr().round(2)
```

```python
print(f"Years: {df["ano"].unique()}")
print()
print(f"UFs: {df["sigla_uf"].unique()}")
```

```python
access_4g_5g_year = df.groupby("ano")["cobertura_pop_4g5g"].mean().reset_index()
print(f"Access 4G/5G per year:\n{access_4g_5g_year}")
```

This sudden decline is explicable. ANATEL changed the methodology to evaluate the access to 4G/5G, they excluded M2M connections and some new criterias like a normalization is now applied.

```python
fiber = df.groupby("ano")['fibra'].sum()
print(fiber)
```

Another decline cause by the change in methodology, where the fiber was diveded into 3 cathegorys and now some registers who were related to those aren't considered fiber anymore

the other decline may happen because of overcounting, very common in the beggining of a new index


```python
estacoes = df.groupby('ano')['adensamento_estacoes'].sum()
print(estacoes)
```

The significant jump in the Station Densification index in 2024 is driven by two factors: real-world infrastructure growth (the high-density requirements of 5G) and a methodological change in ANATEL's calculation. Due to this revision, we cannot use the index to quantify the exact magnitude of the growth compared to previous years, but it reliably indicates the accelerated technological evolution underway.

```python
hhi = df.groupby('ano')['hhi_smp'].sum()
print(hhi)
```

From 2021 to 2023, the data shows aggressive expansion and high penetration (densidade_smp), confirming that mobile access is widely distributed across the Brazilian population. The subsequent sharp drop in 2024 is primarily a result of ANATEL's methodological changes, not a real-world decline in access.

```python
density_smp = df.groupby('ano')['densidade_smp'].sum()
print(density_smp)
```

Increased Competition: The significant drop in the HHI index in 2024 is a strong indicator of an increase in market competition. This is crucial for access distribution, as heightened competition drives better coverage and service quality for consumers.
